```python
import stellarglobe.window
from stellarglobe import layers
window = stellarglobe.window.Window()

show_hips = True

import numpy

def radec2xyz(ra, dec):
    return (
        numpy.cos(dec) * numpy.cos(ra),
        numpy.cos(dec) * numpy.sin(ra),
        numpy.sin(dec),
    )

xyz = radec2xyz(*numpy.random.uniform(-0.1, 0.1, (2, 100)))
```

```python
import colorsys

marker_types = layers.ClickableMarkerLayer.marker_types

markers = [
    dict(
        position=[x, y, z],
        color=[*colorsys.hsv_to_rgb(numpy.random.uniform(), 1, 1), 0.75],
        type=marker_types[i % len(marker_types)]
    )
    for i, (x, y, z) in enumerate(zip(*xyz))    
]
```

```python
from ipywidgets import widgets
from IPython.display import display

out = widgets.Output()
display(out)

def onClick(ev):
    with out:
        print(ev['index'])
```

```python
show_hips = not show_hips

window.set_state(
    layers=[
        layers.TractTileLayer(
            baseUrl=layers.TractTileLayer.pdr3_wide_url,
            visible=not show_hips,
        ),
        layers.TractTileLayer(
            baseUrl=layers.TractTileLayer.pdr3_dud_url,
            visible=not show_hips,
        ),
        # layers.TractTileLayer(jjj
        layers.HipsSimpleLayer(
            visible=show_hips,
            baseUrl='//alasky.cds.unistra.fr/Pan-STARRS/DR1/color-i-r-g',
        ),        
        layers.ConstellationLayer(showNames=False),
        layers.EsoMilkyWayLayer(visible=not show_hips),
        layers.GridLayer(),
        layers.HipparcosCatalogLayer(),
        layers.ClickableMarkerLayer(
            markers=markers,
            onClick=onClick,
        ),        
        # layers.TextLayer(
        #     defaultFont='8pt sans-serif',
        #     defaultColor='white',
        #     texts=[
        #         dict(text=f'{i}', position=[x, y, z])
        #         for i, (x, y, z) in enumerate(zip(*xyz))
        #     ],
        # ),
    ],
)
```
