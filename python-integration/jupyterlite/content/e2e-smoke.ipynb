```python
import math
import time

from hscmap import Window

w = Window(title='e2e smoke')
time.sleep(0.5)
w.jump_to(150, 2, fov=1, duration=0.2)
time.sleep(0.5)
w.sync(force=True)

center = w.camera.center
fov = w.camera.fov
snapshot = w.snapshot_bytes()

print(f'window_title={w.title}')
print(f'center={center}')
print(f'fov={fov}')
print(f'center_ok={math.isclose(center[0], 150, abs_tol=0.1) and math.isclose(center[1], 2, abs_tol=0.1)}')
print(f'fov_ok={math.isclose(fov, 1, abs_tol=0.1)}')
print(f'snapshot_prefix={snapshot[:8]!r}')
print(f'snapshot_is_png={snapshot.startswith(bytes.fromhex("89504e470d0a1a0a"))}')
print(f'snapshot_size={len(snapshot)}')
```
